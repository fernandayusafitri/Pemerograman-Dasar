```python
#Fernanda Ayu Safitri
#4232501010
#25 September 2026
#tujuan: menghitung dan menampilkan hasil operasi aritmatika berupa penjumlahan(+), pengurangan(-), perkalian(*), pembagian(/), pembagian bulat(//), dan modulus(%) menggunakan dua bilangan bulat.
# B. Soal Praktikum
# B.1 Soal Umum

#Soal 1
bilangan1 = 10
bilangan2 = 3

print("Bilangan pertama:", bilangan1)
print("Bilangan kedua:", bilangan2)
print("Hasil penjumlahan (+):", bilangan1 + bilangan2)
print("Hasil pengurangan (-):", bilangan1 - bilangan2)
print("Hasil perkalian (*):", bilangan1 * bilangan2)
print("Hasil pembagian (/):", bilangan1 / bilangan2)
print("Hasil pembagian bulat (//):", bilangan1 // bilangan2)
print("Hasil modulus (%):", bilangan1 % bilangan2)
```

```python
#Fernanda Ayu Safitri
#4232501010
#25 September 2026
#tujuan: menghitung dan menampilkan nilai efisiensi termal berdasarkan energi input dan energi output menggunakan operasi aritmatika.
# B Soal Praktikum
# B.2 Soal Konteks Energi

#Soal 4
energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))
efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal = {efisiensi:.2f}%")
```
