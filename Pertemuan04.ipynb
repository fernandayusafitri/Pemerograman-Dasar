```python
# latihan 1

tahun = int(input("Masukkan tahun: "))
tahun_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

print("Tahun kabisat: ", tahun_kabisat)

# latihan 2

daya = float(input("Masukkan daya peralatan (Watt): "))
waktu = float(input("Masukkan waktu penggunaan (jam): "))
tarif = float(input("Masukkan tarif listrik (Rp/kWh): "))
energi = (daya / 1000) * waktu
biaya = energi * tarif

print("Energi yang digunakan:", energi, "kWh")
print("Biaya listrik: Rp", biaya)
```
