```python
#soal 1 : belanja buku

jumlah_buku = 7
harga_buku = 18500
diskon = 10
uang_budi = 150000
jumlah_orang = 4

total_belanja = jumlah_buku * harga_buku
besar_diskon = total_belanja * diskon // 100
total_bayar = total_belanja - besar_diskon

kembalian = uang_budi - total_bayar

iuran_per_orang = total_bayar // jumlah_orang
sisa = total_bayar % jumlah_orang

uang_cukup = uang_budi >= total_bayar

print("Total belanja: Rp", total_belanja)
print("Besar diskon: Rp", besar_diskon)
print("Total yang harus dibayar: Rp", total_bayar)
print("Uang kembalian: Rp", kembalian)
print("Iuran per orang: Rp", iuran_per_orang)
print("Sisa pembagian: Rp", sisa)
print("Apakah uang Budi cukup?", uang_cukup)
```

```python
#soal 2 : plts atap rumah

jumlah_panel = 12
daya_panel = 450
jam_penyinaran = 5
efisiensi = 0.8
kebutuhan_harian = 25
tarif_listrik = 1444.70

total_daya = jumlah_panel * daya_panel
daya_kw = total_daya / 1000

energi_harian = daya_kw * jam_penyinaran * efisiensi
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

kekurangan_energi = kebutuhan_harian - energi_harian
plts_cukup = energi_harian >= kebutuhan_harian

print("Energi per hari:", energi_harian, "kWh")
print("Energi per bulan:", energi_bulanan, "kWh")
print("Nilai listrik per bulan: Rp", nilai_listrik)
print("Kekurangan energi per hari:", kekurangan_energi, "kWh")
print("Apakah PLTS mencukupi?", plts_cukup)
```

```python
#soal 3 : generator diesel dan rugi kabel

daya_generator = 200
jam_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

energi_harian = daya_generator * jam_operasi

solar_harian = energi_harian * konsumsi_solar
biaya_solar = solar_harian * harga_solar

hari_penuh = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

rugi_daya = arus ** 2 * hambatan

rugi_tinggi = (rugi_daya > 150) and (arus >= 20)

print("Energi per hari:", energi_harian, "kWh")
print("Kebutuhan solar per hari:", solar_harian, "liter")
print("Biaya solar per hari: Rp", biaya_solar)
print("Tangki cukup untuk:", hari_penuh, "hari penuh")
print("Sisa solar:", sisa_solar, "liter")
print("Rugi daya kabel:", rugi_daya, "W")
print("Apakah rugi kabel tergolong tinggi?", rugi_tinggi)
```
