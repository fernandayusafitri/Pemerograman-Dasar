```python
#Soal 2. Energi Harian dan Capacity Factor PLTD

daya_kw = 450
waktu_beroperasi = 8
energi_aktual = 7200
kapasitas_kw = 400

energi_harian = daya_kw * waktu_beroperasi
print("Energi Harian:",energi_harian, "kWh")

capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity Factor: {capacity_factor:.2f} %")
```

Penjelasan program soal 2:
program menghitung energi listrik harian dan capacity factor PLTD. daya 450 kw dikalikan dengan waktu operasi selama 8 jam sehingga menghasilkan energi harian sebesar 3.600 kWh dengan perintah energi_harian = daya_kw * waktu_beroperasi. lalu program hitung capacity factor dari energi aktual dibagi kapasitas kw dikali kan dengan waktu 24 jam lalu dikalikan 100 menghasilkan 75% dengan perintah capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100. soal meminta hasil 2 angka desimal maka menggunakan f-string, yaitu print(f"Capacity Factor: {capacity_factor:.2f} %").

```python
#Soal 6. Status Operasi Boiler

suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")
```

Penjelasan program soal 6:
pakai float(input)) untuk masukin nilai suhu dan nilai tekanan. lalu if suhu >= 550 or tekanan >=14 dipakai untuk memeriksa apakah suhu sampai di angka 550 atau lebih atau tekanan sampai di angka 14 atau lebih, jika salah satu kondisi nya tercapai maka status nya akan menjadi TRIP. elif suhu > 520 and tekanan > 12 dipakai untuk memeriksa apakah nilai suhu lebih dari 520 dan nilai tekanan lebih dari 12 secara bersamaan. jika kedua kondisi terpenuhi, status menjadi WARNING. dan terakhir else digunakan jika kedua kondisi diatas (if & elif) tidak terpenuhi, maka status nya akan NORMAL.

```python
#Soal 8. Pengosongan Tangki Bahan Bakar

volume = 100

while volume > 0:
    volume -= 20
    print("Volume:", volume, "liter")

    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")
```

Penjelasan program soal 8:
variabel volume = 100 untuk nentuin volume awal tangki. while volume > 0 dipakai untuk perulangan yang terus berjalan selama volume masih lebih dari 0. volume -= dipakai untuk mengurangi volume sebanyak 20 liter tiap perulangan. print untuk tampilin volume setelah dikurangi. if volume == dipakai untuk periksa apakah volume udah sampai 20 liter, jika sudah maka akan menampilkan WARNING: Hampir kosong!. jika volume udah sampai di 0 liter, maka perulangan berhenti dan menampilkan TANGKI KOSONG - Pengosongan berhenti.

```python
#Soal 10. Pembacaan Suhu Bearing dengan Sensor Rusak

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1

        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

print("Jumlah data valid:", jumlah_valid)
```

Penjelasan program soal 10:
variabel data adalah daftar suhu bearing di lima pengukuran. variabel jumlah_valid = 0 dipakai untuk hitung jumlah data valid yang udah diperiksa. for i, nilai in enumerate(data, start 1) dipakai untuk periksa setiap data dengan dimulai dari titik 1. try dipakai untuk ngubah data teks jadi angka desimal menggunakan float(). jumlah_valid += 1 dipakai untuk nambahin jumlah data setiap berhasil diubah jadi angka. if suhu > 70.0 dipakai untuk periksa apakah suhu lebih dari 70 derajat celcius. jika iya, akan menampilkan peringatan. except ValueError dipakai untuk menangani data yang tidak terubah jadi angka, contohnya yang "N/A". break dipakai untuk berhentiin perulangan ketika ditemukan data tidak valid.
